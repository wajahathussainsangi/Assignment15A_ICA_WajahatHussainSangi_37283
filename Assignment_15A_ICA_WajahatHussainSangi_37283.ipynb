```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('ride_hailing_dataset.csv')
```

### Q1: What is the average fare by service type?

```python
q1 = df.groupby('service_type')['fare'].mean()
print(q1)
```

**Answer:**
* **Careem:** 544.98
* **InDrive:** 573.51
* **Rickshaw:** 486.64
* **Uber:** 486.13

### Q2: Which city has the most trips?

```python
q2 = df['city'].value_counts()
print(q2)
```

**Answer:**
Islamabad has the most trips (29 trips).

### Q3: What is the average distance per trip?

```python
q3 = df['distance_km'].mean()
print(q3)
```

**Answer:**
The average distance per trip is 9.61 km.

### Q4: Compare trip ratings by gender.

```python
q4 = df.groupby('gender')['trip_rating'].mean()
print(q4)
```

**Answer:**
* **Female:** 2.82
* **Male:** 3.33

Male riders have a higher average trip rating than female riders.

### Q5: Make a bar chart of service type vs. average fare.

```python
q1.plot(kind='bar')
plt.xlabel('Service Type')
plt.ylabel('Average Fare')
plt.title('Service Type vs. Average Fare')
plt.show()
```

**Answer:**
The bar chart shows that InDrive has the highest average fare (573.51), followed by Careem (544.98), while Rickshaw (486.64) and Uber (486.13) have the lowest average fares.
